```python
# Program to count vowels in a string

# Take input from user
text = input("Enter a string: ")

# Define vowels
vowels = "aeiouAEIOU"

# Initialize counter
count = 0

# Loop through each character in the string
for char in text:
    if char in vowels:
        count += 1

# Print the result
print("Number of vowels:", count)
```
